Wedding Planning Portal: 
The Wedding Planning Portal is a Python-based project built to help customers plan and organize their weddings. The portal lets customers sign up, log in, check out wedding services, build a wedding plan, pick services based on their budget and create a wedding report.
The system uses Python functions, loops and conditional statements to handle tasks. It uses dictionaries to store information and file handling to save data. Customer details are saved in a JSON file. This way the data stays safe when the program is not running.

```python
import json
```

Importing Required Module: 
We import the JSON module so that we can store and retrieve customer information from a file.
The JSON module lets the program save Python data into a JSON file and then load that data again when the program starts.

```python
print("**********************************")
print("   WELCOME TO WEDDING PLANNER     ")
print("**********************************")
print(" Plan your dream wedding with us!")
```

Welcome Screen: 
This area shows the message for the Wedding Planning Portal.
It introduces the system to the user and offers a warm welcome before the user explores the available features.

```python
def save_users():
    with open("wedding_users.json", "w") as file:
        json.dump(users, file)


def load_users():
    try:
        with open("wedding_users.json", "r") as file:
            return json.load(file)

    except FileNotFoundError:
        return []
    finally:
        if file is not None:
            file.close()
```

File Handling: 
This section contains two functions: save_users() and load_users().
save_users() saves the registered users and their wedding information into a JSON file.
load_users() retrieves saved user information when the program starts.
The try and except statements are used to handle the situation where the JSON file does not exist. If the file is not found load_users() returns a list instead of stopping the program.
This allows customer information to remain available even after the program is closed.

```python
users = load_users()
current_user = None
services = {"Photography": 25000, "Catering": 50000, "Decoration": 30000, "Makeup": 20000, "Videography": 35000}
```

Initializing Wedding Services: 
This section initializes the variables that are used throughout the program.
users holds the registered customer information loaded from the JSON file.
current_user holds the information of the customer who is currently logged in.
services is a dictionary that contains the wedding services and their prices.
The available wedding services include photography, catering, decoration, makeup and videography.

```python
def register_user():
    print("*********** CUSTOMER REGISTRATION ***********")
    name = input("Enter your name: ")
    email = input("Enter your email: ")
    for user in users:
        if user["email"] == email:
            print("This email is already registered!")
            return
    password = input("Create a password: ")
    user = {"name": name,"email": email,"password": password}
    users.append(user)
    save_users()
    print("Registration successful!")
    return user
```

Customer Registration: 
This part lets new customers make an account on the Wedding Planning Portal.
The customer types in their name, email address and a password. This data is saved in a dictionary. Then added to the list of users.
The program checks if the email address is already, in use. If the email is already registered the program shows a message. Stops the registration from being completed.
Once the registration is successful the updated user details are saved in the JSON file.

```python
def user_login():
    global current_user
    print("*********** CUSTOMER LOGIN ***********")
    email = input("Enter your email: ")
    password = input("Enter your password: ")
    for user in users:
        if user["email"] == email and user["password"] == password:
            current_user = user
            print("Login successful!")
            return
    print("Invalid email or password.")
```

Customer Login: 
This section allows registered customers to log in to their accounts.
The program asks customer to enter email address and password. Then program uses a for loop to check users.
If entered information matches an existing account customer is logged in successfully. Customer information is stored in current_user variable.
If information does not match, program displays invalid login message.

```python
def view_services():
    print("*********** WEDDING SERVICES ***********")
    for service in services:
        print(service, "Rs.", services[service])
    return services
```

Viewing Wedding Services: 
This part shows all the wedding services that can be found on the portal.
A for loop is used to go through the services dictionary and show each service along, with its price.
People who want to get married can look at the options before they start making their wedding plans.

```python
def plan_wedding():
    if current_user == None:
        print("Please login first.")
        return
    print("*********** PLAN YOUR WEDDING ***********")
    bride = input("Enter bride's name: ")
    groom = input("Enter groom's name: ")
    date = input("Enter wedding date: ")
    location = input("Enter wedding location: ")
    try:
        guests = int(input("Enter number of guests: "))
        budget = int(input("Enter your budget: "))
    except ValueError:
        print("Invalid input! Please enter numbers only.")
        return
    wedding = {"bride": bride,"groom": groom,"date": date,"location": location,"guests": guests,"budget": budget }
    current_user["wedding"] = wedding
    save_users()
    print("Wedding plan created successfully!")
    return wedding
```

Creating a Wedding Plan: 
This section lets a person who is signed in make their wedding plan.
The person gives the name of the bride the name of the groom the date of the wedding the place where it will happen how many people will come and how much money is available for the event.
All of this information is kept in a dictionary named wedding. Is put into the account of the current person.
The program also uses try and except to deal with numbers that're not correct, for the number of guests and the budget.
Once the wedding plan is made correctly the new information is written into the JSON file.

```python
def view_wedding_plan():
    if current_user == None:
        print("Please login first.")
        return
    if "wedding" not in current_user:
        print("No wedding plan found.")
        return
    wedding = current_user["wedding"]
    print("*********** YOUR WEDDING PLAN ***********")
    print("Bride:", wedding["bride"])
    print("Groom:", wedding["groom"])
    print("Date:", wedding["date"])
    print("Location:", wedding["location"])
    print("Guests:", wedding["guests"])
    print("Budget: Rs.", wedding["budget"])
```

Viewing the Wedding Plan: 
This section lets customers see the wedding plans they have made before.
First the program checks if a customer is logged in. Then the program checks if a wedding plan is stored in the customer’s account.
If a wedding plan is available the wedding plan shows the wedding details. The wedding plan shows the bride’s name, the groom’s name, the date the location the number of guests and the budget.
If no wedding plan exists the program shows a message.

```python
def select_services():
    if current_user == None:
        print("Please login first.")
        return
    if "wedding" not in current_user:
        print("Please create a wedding plan first.")
        return
    selected_services = {}
    print("*********** SELECT WEDDING SERVICES ***********")
    while True:
        print("Available Services:")
        for service in services:
            print(service, "- Rs.", services[service])
        choice = input("Enter service name or type done: ")
        if choice.lower() == "done":
            break
        elif choice in services:
            if choice in selected_services:
                print("Service already selected.")
            else:
                selected_services[choice] = services[choice]
                print(choice, "added successfully!")
        else:
            print("Invalid service name. Please try again.")

    current_user["wedding"]["selected_services"] = selected_services

    total = sum(selected_services.values())

    current_user["wedding"]["service_total"] = total

    print("Your selected services:")

    for service in selected_services:
        print(service, " Rs.", selected_services[service])

    print("Total service cost: Rs.", total)
    if total <= current_user["wedding"]["budget"]:
        print("Your selected services are within your budget!")
    else:
        print("Your selected services exceed your budget!")
    save_users()
    return selected_services
```

Selecting Wedding Services: 
This section lets customers pick the wedding services that they want to add to their wedding plan. 
The program shows the services and their prices. A while loop lets customers choose services until they type the word 'done'. 
A dictionary holds the selected services and their prices. The program stops customers from choosing the service more, than once. 
The sum() function adds up the cost of the selected services. This total is compared with the customers wedding budget.
Finally the selected services and their total cost are saved in the customers wedding plan. Written to the JSON file. 

```python
def generate_report():
    if current_user == None:
        print("Please login first.")
        return
    if "wedding" not in current_user:
        print("Please create a wedding plan first.")
        return
    wedding = current_user["wedding"]
    print("*********** WEDDING REPORT ***********")
    print("Customer:", current_user["name"])
    print("Bride:", wedding["bride"])
    print("Groom:", wedding["groom"])
    print("Date:", wedding["date"])
    print("Location:", wedding["location"])
    print("Guests:", wedding["guests"])
    print("Budget: Rs.", wedding["budget"])
    if "selected_services" in wedding:
        print("Selected Services:")
        for service in wedding["selected_services"]:
            print(service, "- Rs.", wedding["selected_services"][service])
        print("Total Service Cost: Rs.", wedding["service_total"])
    return wedding
```

Generating a Wedding Report: 
This section creates a summary of the customers wedding plan.
The program makes sure the customer is logged in and that a wedding plan has been set up.
The report shows the customers name, the brides name, the grooms name, the wedding date, the place the number of guests and the budget.
If the customer has chosen wedding services the report also shows the services that were chosen and the total cost of those services.

```python
while True:
    print("*********** WEDDING PLANNING PORTAL ***********")

    print("1. Register")
    print("2. Login")
    print("3. View Wedding Services")
    print("4. Plan My Wedding")
    print("5. View My Wedding Plan")
    print("6. Generate Report")
    print("7. Select Services")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        register_user()

    elif choice == "2":
        user_login()

    elif choice == "3":
        view_services()

    elif choice == "4":
        plan_wedding()

    elif choice == "5":
        view_wedding_plan()

    elif choice == "6":
        generate_report()

    elif choice == "7":
        select_services()

    elif choice == "8":
        print("Thank you for using Wedding Planner!")
        break

    else:
        print("Invalid choice. Please try again.")
```
